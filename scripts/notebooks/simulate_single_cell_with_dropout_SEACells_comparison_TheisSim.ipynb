```python
def get_obs_data(h5ad_file_path):
    adata = ad.read_h5ad(h5ad_file_path, backed='r')
    obs_df = adata.obs.copy()
    adata.file.close()
    return obs_df

def get_model_data(model_path):
    with open(model_path, "rb") as f:
        model = pickle.load(f)
        assignments = model["model"].get_hard_assignments()
    return assignments



def calc_celltype_purity(obs_df, assignments):
    obs_df = obs_df.copy()
    obs_df['SEACell'] = assignments
    
    def purity_func(group):
        # Get the counts of each Group in the SEACell
        counts = group['Group'].value_counts()
        # Identify the most common Group and its count
        most_common_group = counts.idxmax()
        most_common_count = counts.max()
        # Total number of cells in the SEACell
        total_count = counts.sum()
        # Calculate purity
        purity = most_common_count / total_count
        return pd.Series({'Group': most_common_group, 'celltype_purity': purity})
    
    # Apply the function to each SEACell group
    result = obs_df.groupby('SEACell').apply(purity_func)
    return result
```

```python
out_merged = "/local/Projects/AAkalin_Neuroblastoma/Results/simulate_single_cell/h5ad/with_dropout/SEACells/merged_seacell_data"
ad_dict = dict()
obs_cols = ['sample_name', 'file_path', 'sim', 'Batch', 'parent_sample',
       'seacell_path', 'lib.loc', 'dropout.type', 'dropout.shape',
       'dropout.mid', 'sparsity']
for ind, row in samplesheet_sim2.iterrows():
    # load seacell adata
    adata = sc.read_h5ad(row.seacell_path)
    # load obs of the original preprocessed dataset
    preprocessed_obs_df = get_obs_data(row.preprocessed_path)
    # load seacell model
    assignments = get_model_data(row.model_path)
    # calculate celltype purity and add it to the obs. Then filter seacells with celltype purity > 0.95
    celltype_purity = calc_celltype_purity(preprocessed_obs_df, assignments)
    adata.obs = adata.obs.join(celltype_purity, how = "inner")
    adata = adata[adata.obs["celltype_purity"] > 0.95]
    
    # add metadata of the simulation to the adata object
    obs_dict = row[obs_cols].to_dict() 
    adata.obs = adata.obs.assign(**obs_dict)
    adata.obs["unique_cell_id"] = adata.obs.apply(lambda x: f"{x.name}_{x.Batch}", axis=1)
    adata.obs_names = adata.obs["unique_cell_id"]
    adata.obs.index = adata.obs["unique_cell_id"]
    
    parent_sample = row.parent_sample
    if not parent_sample in ad_dict.keys():
        ad_dict[parent_sample] = []
    ad_dict[parent_sample].append(adata)

def calculate_sparsity(adata):
    def count_zeros(arr, tol=1e-10):
        return np.sum(np.abs(arr) < tol)
    sparsity = count_zeros(adata.X.toarray()) / float(adata.X.size)
    sparsity = np.round(sparsity, 4) * 100
    return sparsity

ad_dict_concat = dict()
for sample, ads in ad_dict.items():
    ad_dict_concat[sample] = ad.concat(ads, axis = 0)
    ad_dict_concat[sample].obs["seacell_sparsity"] = calculate_sparsity(ad_dict_concat[sample])
    ad_dict_concat[sample].write_h5ad(join(out_merged, f"{sample}_merged_seacell.h5ad"))

    
```
